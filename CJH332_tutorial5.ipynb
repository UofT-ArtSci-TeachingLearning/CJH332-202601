#### Filtering data using conditional operators

One of the most useful things you can do with a DataFrame is filter it to keep only the rows that meet a certain condition. This works a lot like filtering data in Excel, but with code.

In pandas, you create a condition (something that is `True` or `False` for each row), and then use it inside square brackets to filter the DataFrame.

Common conditional operators in pandas

| Operator | Meaning                  | Example                             | Reads as…              |
| -------- | ------------------------ | ----------------------------------- | ---------------------- |
| `==`     | equals                   | `df["x"] == 30`                     | x is equal to 30       |
| `!=`     | not equal                | `df["x"] != 30`                     | x is not 30            |
| `>`      | greater than             | `df["x"] > 80`                      | x is more than 80      |
| `<`      | less than                | `df["x"] < 18`                      | x is less than 18      |
| `>=`     | greater than or equal to | `df["x"] >= 50000`                  | x is at least 50,000   |
| `<=`     | less than or equal to    | `df["x"] <= 4`                      | x is 4 or below        |
| `&`      | and (combine conditions) | `(df["x"] > 18) & (df["y"] == 30)`  | x > 18 **and** y is 30 |
| `\|`     | or (combine conditions)  | `(df["x"] < 18) \| (df["y"] == 30)` | x < 18 **or** y is 30  |


```python
# Filters the DataFrame to include rows with year >= 2010
bca[bca["year"] >= 2010]
```

```python
# Filters the DataFrame to rows from 2010 onward AND with rate_cases1 greater than 6
bca[(bca["year"] >= 2010) & (bca["rate_cases1"] > 6)]
```

#### Conditional Logic (if / elif / else)

Conditional logic allows your program to make decisions.
Instead of running every line of code, Python can choose what to do based on whether a condition is True or False.

##### Basic **if** statement

Runs a block of code **only if** the condition is true.

```python
if bca["rate_cases1"].mean() < 6:
    print("rate cases is less than 6") # Runs if mean is less than 6
```

##### **if / else**

else runs when the condition in the if is **not true**.

```python
if bca["death_rate"].mean() > 8: 
    print("death cases is greater than 8") # Runs if the mean is greater than 8
else: 
    print("death cases is 8 or below") # Runs for all other cases (mean =< 8)
```

##### **elif** (else if)

elif lets you check multiple conditions in order.

Python will run the first condition that is true and ignore the rest.

```python
# Check the mean of pct_survival and compare it to 40
if bca["pct_survival"].mean() == 40:
    print("pct survival rate is 40")              # Runs if the mean equals 40

elif bca["pct_survival"].mean() < 40:
    print("pct survival rate is less than 40")    # Runs if the mean is below 40

else:
    print("pct survival rate is greater than 40") # Runs for all other cases (mean > 40)
```

### while loops

In an earlier tutorial we learned about for loops, and now that we have better sense of logical operators in Python, we can also use while loops. while loops are best for cases when you want to run a loop until a condition is met.

```python
myNumbers = []
counter = 0

while counter < 5:
    print(counter)
    myNumbers.append(counter)
    counter = counter + 1
    
print(myNumbers)
```
