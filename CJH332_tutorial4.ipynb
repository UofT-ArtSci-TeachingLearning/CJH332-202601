#### What are libraries?

So far, we’ve been using Python’s built-in functions. But what if we want to do more?

Libraries are collections of pre-written code that add new tools to Python. One of the biggest advantages of Python is the size of the community that uses it, and this community has created a significant collection of libraries that extends the functionality of the language. 

By **importing** a library, you instantly get access to extra features—like working with data, making plots, or doing more advanced math—without having to write everything yourself.

### What is pandas?
`pandas` is a Python library that makes it easy to work with data.

pandas gives you:

DataFrames — tables with rows and columns, similar to spreadsheets

Tools to explore and manipulate data — sorting, filtering, cleaning, summarizing

Pandas is one of the most important tools for doing data analysis in Python!

```python
import pandas as pd # pd is the conventional alias for pandas
```

#### DataFrames

A DataFrame is a table made up of rows and columns—similar to a spreadsheet! Each column has a name, each row has an index, and together they form a structure that’s easy to explore and analyze.

One of the most common ways to create a DataFrame is by loading data from a CSV file.

Here’s how you load a CSV into a DataFrame:

```python
bca = pd.read_csv("Data/BrainCancerData.csv")
bca
```

#### Using Relative Path vs Absolute Path
<b>Relative paths</b> are often used to find data files relative to the location of the notebook.

In the code above, we used a relative path.

Explanation of <code>"../Data/BrainCancerData.csv"</code>:
<ul>
    <code>..</code> means it is one folder up
    <code>Data</code> is the folder that contains the data files
    <code>BrainCancerData.csv</code> is the name of the file that contains the data
</ul>

<b>Note</b>: It is a good practice to follow the capitalization of the folder name and filename

<b>Absolute paths</b> are used to find the file in a specific location, regardless of where the notebook file exists.

This requires the entire file path.
<ul>
    In Windows, the path often starts with <code>C:/Users/...</code>
    In Mac, the path often starts with <code>/Users/...</code>
</ul>

#### Accessing Columns

Once you have a DataFrame, you’ll often want to look at specific columns. pandas provides simple and powerful tools for this.

You can select a single column using its name:

```python
bca["year"]
```

To select multiple columns, provide a list of column names:

```python
bca[["rate_cases1", "rate_cases2"]]
```

#### Basic Summary Methods

We can get a quick overview of a DataFrame by using the `info()` method.

info() prints a summary that helps us understand the structure of our data. It includes:

- The shape — how many rows and columns the DataFrame has

- Column names and their data types (dtypes)

- How many non-null (non-missing) values are in each column

```python
bca.info()
```

*Recall:* "int" (integer) is a numeric value that is a whole number while a float represents a numeric value that has decimal places. Don't worry about the '64' beside the float and int

Pandas also makes it easy to calculate basic summary statistics for numeric columns!

**Mean**

The mean (average) of a column:

```python
bca["rate_cases1"].mean()
```

**Standard deviation (sd)**

The standard deviation, which measures how spread out the values are:

```python
bca["rate_cases1"].std()
```
