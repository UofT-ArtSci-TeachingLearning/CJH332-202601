Let’s create a scatter plot to visualize the relationship between the predictor variable (**year**) on the x-axis, and we place the response variable that we want to predict (**PCT survival**) on the y-axis.

```python
# Import the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load in the bca DataFrame
bca = pd.read_csv("Data/BrainCancerData.csv")
```

```python
fig, ax = plt.subplots()
ax.scatter(bca["year"], bca["pct_survival"])
ax.set_xlabel("Year")
ax.set_ylabel("PCT Survival")
ax.set_title("Year vs. PCT Survival")
plt.show()
```

We can see that in this data set as the year increases PCT survival rates tend to increase as well. This means we might be able to predict the PCT survival of an individual we haven't seen yet based on the year.

Now that we’ve visualized at our data, let’s quantify the relationship.

#### Simple Linear Regression with statsmodels

`statsmodels` is a Python library designed for statistical modeling

```python
import statsmodels.formula.api as smf
```

Simple linear regression is a method to find the best straight line that shows the relationship between two things. It’s called "linear" because it shows a straight-line relationship between the two things.

The equation for the straight line is:

$$
\text{PCT Survival} = b_0 + b_1 \times (\text{year})
$$

where:

- $ b_0 $ is the PCT survival rate when the year is 0 (the intercept).
- $ b_1 $ is how much the PCT survival rate changes for each additional year (the slope).

statsmodels provides a formula interface that allows you to specify models in a familiar formula syntax (similar to R), which makes it easy to specify the model and interact with your data.

Syntax:

> model(response variable ~ predictor, data).fit()

```python
# Simple linear regression formula
model = smf.ols("pct_survival ~ year", data=bca).fit()
```

*Wait, what is OLS?*

Ordinary Least Squares (OLS) is just a method for drawing the best-fitting straight line through your data.
It chooses the line that gets as close as possible to all the points by making the overall “errors” as small as they can be.

Think of it like trying to sketch a line that best matches the trend in a scatter plot — OLS finds the line that fits the data the most closely.

Once we have our model, we can use the  `summary` function to tell us important things like:

- R-squared (how much of the data the model explains)

- p-values (which variables are statistically significant)

- F-statistic (overall model fit)

- t-statistics (significance of individual coefficients)

```python
print(model.summary())
```

Interpreting the Slope

> The slope is the coefficient for the variable year. In our model, the slope is 0.3575, meaning: For each additional year, pct_survival increases by about 0.36 on average. This helps us connect the model to the real world: survival rates have been improving over time. Because the p-value < 0.05, the increase is statistically significant — the trend we observe is very unlikely to be due to random chance.


Interpreting t-values

> In linear regression, the null hypothesis is that each coefficient (intercept or slope) is 0 — meaning the variable has no effect. A large |t-value| suggests the coefficient is meaningfully different from zero. Both the intercept and the year coefficient have large t-values, which supports rejecting the null hypothesis. This means year is truly related to pct_survival — not just noise.

Interpreting R-squared

> An R-squared of 0.824 means the model explains about 82% of the variation in pct_survival.This is a very strong relationship for a simple linear regression with only one predictor.

Interpreting the F-statistic

> The F-statistic = 187.2 with a tiny p-value (1.12e-16) tells us the overall model is highly significant. Together, the predictor explain much more variation than a model with no predictors. In short: the regression model fits the data extremely well.
