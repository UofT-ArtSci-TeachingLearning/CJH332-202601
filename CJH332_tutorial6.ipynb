pandas makes it easy to organize, summarize, and clean your data.
Here are some of the most common operations you’ll use when analyzing datasets.

```python
import pandas as pd
```

```python
bca = pd.read_csv("Data/BrainCancerData.csv")
```

##### **Renaming Columns**

Use `.rename()` to rename one or more columns.

```python
# Rename one column
bca = bca.rename(columns={
    "rate_cases1": "rate_cases_pre",
    "rate_cases2": "rate_cases_post"
})
bca
```

Renaming helps make column names shorter, clearer, or easier to interpret!

##### Grouping Data

Use `.groupby()` when you want to group rows that share a value in a column
(e.g., by year, by city, by gender, etc.).

Example: group by "new_treatment"

```python
bca.groupby("new_treatment") #group by treatment category
```

On its own, this doesn’t show the results yet.
To actually see something useful, you need to summarize the groups.

##### Summarizing Data (Aggregations)

After grouping, you can compute summary statistics (mean, sum, count, etc.).

```python
# group the DataFrame by 'new_treatment' and find the mean of rate_cases_pre within each group
bca.groupby("new_treatment")["rate_cases_pre"].mean()
```

We can also summaries on multiple columns!

```python
bca.groupby("new_treatment").agg({
    "rate_cases_pre": "mean",
    "rate_cases_post": "mean"
})
```

What this code does:

`bca.groupby("new_treatment")`
Groups the DataFrame by the values in the "new_treatment" column.
Each treatment group becomes its own subgroup.

`.agg({...})`
The .agg() function lets you apply multiple summary statistics to one or more columns.
inside the {}, this tells pandas:

 - Take the column "rate_cases_pre" and compute its mean within each treatment group
 - Take the column "rate_cases_post" and compute its mean within each treatment group
