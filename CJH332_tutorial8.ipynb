#### Common Plot Types in Data Visualization

| Plot Type        | What It Shows                                       | Best Used For                                                  |
| ---------------- | --------------------------------------------------- | -------------------------------------------------------------- |
| **Line Plot**    | Continuous data connected over an axis (often time) | Trends, changes over time, continuous measurements             |
| **Scatter Plot** | Individual data points on two axes                  | Relationships between two numeric variables; correlation       |
| **Bar Chart**    | Heights/lengths representing category values        | Comparing categories or groups                                 |
| **Histogram**    | Data grouped into bins                              | Distribution of a single numeric variable                      |
| **Boxplot**      | Median, quartiles, and outliers                     | Comparing distributions across groups                          |
| **Pie Chart**    | Parts of a whole                                    | Simple proportion comparisons (best with few categories)       |

#### matplotlib

matplotlib is the foundational data visualization library in Python.


You can use matplotlib to customize colors, labels, titles, axes, sizes, and much more

Many advanced visualization libraries beyond the scope of this tutorial (like seaborn) extend or enhance matplotlib’s plotting features

##### **pyplot-style plotting**

pyplot-style plotting is convenient for quick, exploratory data visualization—especially when you don’t need a lot of customization. 


```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load in the bca DataFrame
bca = pd.read_csv("Data/BrainCancerData.csv")
```

Here’s an example using the pyplot style:

```python
# plot year on the x-axis and rate_cases_pre on the y-axis to see how the rate changes over time
fig, ax = plt.subplots()
ax.plot(bca['year'], bca['rate_cases1'])
plt.show()
```

By default, plt.subplots() create line plots.

The basic plot isn’t very informative, so let’s add context with labels and a title.
To do this, we can call methods on the current Axes object:

```python
# Create a simple line plot of rate_cases_pre over the years
fig, ax = plt.subplots()
ax.plot(bca['year'], bca['rate_cases1'])

# Add a title and axis labels for context
ax.set_title("Rate Cases Pre-Treatment vs. Year", fontsize=14)
ax.set_xlabel("Year")
ax.set_ylabel("Rate Cases Pre-Treatment")

# Add a translucent grid to make values easier to read
ax.grid(alpha=0.3)

# Ensure the plotted points/line appear above the grid lines
ax.set_axisbelow(True)

# Display the final plot
plt.show()
```

`ax.set_title()` sets the title

`ax.set_xlabel() / ax.set_ylabel()` set axis labels

`ax.grid()` adds a grid (with optional transparency via `alpha`)

`ax.set_axisbelow(True)` ensures the grid draws behind the markers

We could also make a scatter plot!

```python
# Create a simple scatter plot of rate_cases_pre over the years
fig, ax = plt.subplots()
ax.scatter(bca['year'], bca['rate_cases1'])

# Add a title and axis labels for context
ax.set_title("Rate Cases Pre-Treatment vs. Year", fontsize=14)
ax.set_xlabel("Year")
ax.set_ylabel("Rate Cases Pre-Treatment")

# Add a translucent grid to make values easier to read
ax.grid(alpha=0.3)

# Ensure the plotted points/line appear above the grid lines
ax.set_axisbelow(True)

# Display the final plot
plt.show()
```
